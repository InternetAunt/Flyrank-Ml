## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

What this work can say: These are observed, historical patterns — which content items show declining engagement/CTR signals, and how strongly those signals correlate with the "declining" label in this dataset. Any model built here would be decision-support: it flags which items look worth an editor's attention, not a verdict.
What this work can never say: This is not causal proof that any particular signal causes decline, nor a guarantee that refreshing a flagged item will fix it. It also can't predict future search-engine behavior in general (e.g. "predicting Google") — only describe patterns in this specific historical window, for these specific clients. Any "prediction" is really a same-time-window classification, not a forecast of what happens after action is taken.
